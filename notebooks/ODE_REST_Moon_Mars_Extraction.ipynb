# ODE extraction for the Moon and Mars: topography, geology, and mineralogy
**NASA PDS Orbital Data Explorer (ODE REST `live2`)** — NASA Space Apps 2026 project

Notebook workflow:

| Step | Purpose | Output |
|---|---|---|
| 1 | Select datasets by research area and celestial body | `app/data/catalogs/selected_datasets.csv` |
| 2 | List variables, units, and sources for each dataset | `app/data/catalogs/dataset_variables.csv` |
| 3 | Validate available instruments and product types against ODE (`query=iipt`) | `app/data/catalogs/ode_validation_{body}.csv` |
| 4 | Plan downloads for products that cover each AOI | `app/data/catalogs/download_plan.csv` |
| 5 | Download source files to `app/data/raw/{moon,mars}/{topography,geology,mineralogy}/…` | Raw files |
| 6 | Process raster data to `app/data/processed/{moon,mars}/…` | CSV files by dataset and AOI |
| 7 | Build a shared cell grid that combines the three categories | `INTEGRATED__{aoi}.csv` |
| 8 | Summarize row counts, columns, valid values, and file sizes | `app/data/catalogs/final_dataset_catalog.csv` |

**Inputs and `app/data/` layout**
- `app/data/input/sites.csv` contains site coordinates. The notebook uses the first five sites for each body.
- `app/data/raw/` stores downloads; `app/data/processed/` stores derived tables; `app/data/catalogs/` stores plans and metadata.
- Only `app/data/processed/topography_sites_25km.csv` is versioned for deployment. `raw/`, `catalogs/`, and `reference/` are excluded from Git.

**Data limitations**
- ODE is a catalog: rasters (DEMs and maps) can be sampled into tables. Images and spectral cubes (CRISM, M3) are recorded as footprints and metadata; cube downloads are optional.
- Images do not contain geologic labels. Geologic units, ages, and crater density come from external maps and catalogs (USGS, Robbins). Step 7 can use them when paths are configured.
- Dataset identifiers (`IHID`, `IID`, `PT`) are starting references. Step 3 validates them against ODE and reports products hosted elsewhere (for example, HRSC and OMEGA data from ESA/PSA).

## Step 3 — ODE REST client and catalog validation
Query `query=iipt` (instrument host / instrument / product type) and check which selected datasets are available in ODE.

```python
ODE_REST_URL = "https://oderest.rsl.wustl.edu/live2"


def crear_sesion():
    s = requests.Session()
    retry = Retry(
        total=4,
        backoff_factor=1.5,
        status_forcelist=[429, 500, 502, 503, 504],
        allowed_methods=["GET"],
    )
    s.mount("https://", HTTPAdapter(max_retries=retry))
    s.headers["User-Agent"] = "ode-space-apps-notebook/1.0"
    return s


SESION = crear_sesion()


def ode_get(params: dict, timeout: int = 90) -> dict:
    """GET a ODE REST; devuelve el bloque ODEResults o lanza error legible."""
    params = {k: v for k, v in params.items() if v is not None}
    params.setdefault("output", "JSON")
    resp = SESION.get(ODE_REST_URL, params=params, timeout=timeout)
    resp.raise_for_status()
    data = resp.json()
    res = data.get("ODEResults", data)
    estado = str(res.get("Status", "")).lower()
    if estado and estado not in ("success", "ok"):
        raise RuntimeError(f"ODE returned an error: {res.get('Error', res)}")
    return res


def _buscar_dicts(obj, claves):
    """Recorre el JSON y devuelve todos los dicts que contengan las claves dadas."""
    out = []
    if isinstance(obj, dict):
        if all(k in obj for k in claves):
            out.append(obj)
        for v in obj.values():
            out += _buscar_dicts(v, claves)
    elif isinstance(obj, list):
        for v in obj:
            out += _buscar_dicts(v, claves)
    return out


_CACHE_IIPT = {}


def listar_iipt(target: str) -> pd.DataFrame:
    """Todos los (IHID, IID, PT) disponibles en ODE para un body."""
    if target not in _CACHE_IIPT:
        res = ode_get({"query": "iipt", "target": target})
        filas = _buscar_dicts(res, ["IHID", "IID", "PT"])
        if not filas:
            raise RuntimeError(
                "IIPT response is missing IHID/IID/PT; check the ODE REST documentation."
            )
        df = (
            pd.DataFrame(filas).drop_duplicates(subset=["IHID", "IID", "PT"]).reset_index(drop=True)
        )
        col_n = next(
            (
                c
                for c in df.columns
                if c.lower() in ("numberproducts", "number_products", "productcount", "count")
            ),
            None,
        )
        df["n_productos"] = pd.to_numeric(df[col_n], errors="coerce") if col_n else np.nan
        _CACHE_IIPT[target] = df
    return _CACHE_IIPT[target]


def _contiene(serie, tokens: str):
    toks = [t for t in tokens.split("|") if t]
    if not toks:
        return pd.Series(False, index=serie.index)
    return serie.astype(str).str.contains(
        "|".join(map(re.escape, toks)), case=False, regex=True, na=False
    )


def candidatos_ode(d: dict) -> pd.DataFrame:
    df = listar_iipt(d["body"])
    ih = df["IHID"].astype(str) + " " + (df["IHName"].astype(str) if "IHName" in df else "")
    ii = df["IID"].astype(str) + " " + (df["IIName"].astype(str) if "IIName" in df else "")
    return df[_contiene(ih, d["ihid"]) & _contiene(ii, d["iid"])]


def elegir_pts(d: dict, cand: pd.DataFrame, max_pts: int = 2) -> list:
    """Elige hasta max_pts tipos de producto: los que cumplen el patrón; si ninguno, el de más productos."""
    if cand.empty:
        return []
    sel = cand[_contiene(cand["PT"], d["pt_pref"])] if d["pt_pref"] else cand.iloc[0:0]
    if sel.empty:
        sel = cand.sort_values("n_productos", ascending=False)
    sel = sel.sort_values("n_productos", ascending=False).head(max_pts)
    return [(r.IHID, r.IID, r.PT) for r in sel.itertuples()]


PLAN_ODE = {}  # ds_id -> [(ihid, iid, pt), ...]


def validar_catalogo(bodies=BODIES) -> pd.DataFrame:
    filas = []
    for d in DATASET_SELECTION:
        if d["body"] not in bodies:
            continue
        if d["access"] == "external":
            filas.append(
                dict(
                    ds_id=d["ds_id"],
                    body=d["body"],
                    instrument=d["instrument"],
                    estado="Externo (fuera de ODE)",
                    ihid_ode="",
                    iid_ode="",
                    pt_elegidos="",
                    pts_disponibles="",
                    n_productos_total=np.nan,
                )
            )
            continue
        try:
            cand = candidatos_ode(d)
        except Exception as e:
            filas.append(
                dict(
                    ds_id=d["ds_id"],
                    body=d["body"],
                    instrument=d["instrument"],
                    estado=f"Error consulta: {e}",
                    ihid_ode="",
                    iid_ode="",
                    pt_elegidos="",
                    pts_disponibles="",
                    n_productos_total=np.nan,
                )
            )
            continue
        pts = elegir_pts(d, cand)
        PLAN_ODE[d["ds_id"]] = pts
        filas.append(
            dict(
                ds_id=d["ds_id"],
                body=d["body"],
                instrument=d["instrument"],
                estado="OK en ODE" if pts else "No encontrado en ODE",
                ihid_ode="|".join(sorted(cand["IHID"].astype(str).unique())),
                iid_ode="|".join(sorted(cand["IID"].astype(str).unique())),
                pt_elegidos="|".join(f"{i}/{j}/{p}" for i, j, p in pts),
                pts_disponibles=", ".join(
                    f"{r.PT}({'?' if pd.isna(r.n_productos) else int(r.n_productos)})"
                    for r in cand.itertuples()
                ),
                n_productos_total=cand["n_productos"].sum(min_count=1),
            )
        )
    return pd.DataFrame(filas)


try:
    df_validacion = validar_catalogo()
    for c in BODIES:
        df_validacion.query("body == @c").to_csv(
            DIR_CATALOGS / f"ode_validation_{c}.csv", index=False, encoding="utf-8-sig"
        )
    display(df_validacion[["ds_id", "instrument", "status", "selected_pts", "total_products"]])
except Exception as e:
    print("✗ Could not validate against ODE (offline?):", e)
```

## Step 4 — Query each AOI and plan downloads
For each active dataset and AOI, find covering products and save their metadata. For rasters, choose one product per AOI: the highest-resolution product below `MAX_FILE_MB`.

```python
def calculate_bounding_box(
    center_lat: float, center_lon: float, lado_km: float = 500.0, radio_km: float = 1737.4
):
    delta_lat = lado_km / (radio_km * (math.pi / 180.0))
    cos_lat = max(math.cos(math.radians(center_lat)), 0.001)
    delta_lon = min(lado_km / (radio_km * cos_lat * (math.pi / 180.0)), 359.0)
    return {
        "minlat": round(max(center_lat - delta_lat / 2.0, -90.0), 5),
        "maxlat": round(min(center_lat + delta_lat / 2.0, 90.0), 5),
        "westernlon": round((center_lon - delta_lon / 2.0) % 360.0, 5),
        "easternlon": round((center_lon + delta_lon / 2.0) % 360.0, 5),
        "delta_lat_deg": delta_lat,
        "delta_lon_deg": delta_lon,
        "center_lat": center_lat,
        "center_lon": center_lon,
        "area_km2": lado_km * lado_km,
        "radio_km": radio_km,
    }


def _a_float(x, default=np.nan):
    try:
        return float(x)
    except (TypeError, ValueError):
        return default


def query_ode_products(
    target,
    ihid=None,
    iid=None,
    pt=None,
    bbox=None,
    pdsid=None,
    limite_total=MAX_METADATA_PRODUCTS,
    pagina=100,
):
    """Devuelve (df_meta, df_archivos). Paginado con limit/offset."""
    base = {"target": target, "query": "product", "results": "pmf"}
    if pdsid:
        base["pdsid"] = pdsid
    else:
        base.update(ihid=ihid, iid=iid, pt=pt)
        if bbox:
            base.update(
                minlat=bbox["minlat"],
                maxlat=bbox["maxlat"],
                westernlon=bbox["westernlon"],
                easternlon=bbox["easternlon"],
            )
    productos, offset = [], 0
    while len(productos) < limite_total:
        lim = min(pagina, limite_total - len(productos))
        res = ode_get(dict(base, limit=lim, offset=offset))
        bloque = res.get("Products", {})
        bloque = bloque.get("Product", []) if isinstance(bloque, dict) else []
        if isinstance(bloque, dict):
            bloque = [bloque]
        productos += bloque
        if len(bloque) < lim:
            break
        offset += len(bloque)

    meta, archivos = [], []
    for p in productos:
        pid = p.get("pdsid") or p.get("Product_name") or p.get("ode_id")
        fila = {k: v for k, v in p.items() if isinstance(v, (str, int, float, bool))}
        fila.update(
            pdsid=pid,
            escala_m_pix=_a_float(p.get("Map_scale")),
            resolucion_pix_deg=_a_float(p.get("Map_resolution")),
        )
        fs = (p.get("Product_files") or {}).get("Product_file", [])
        fs = [fs] if isinstance(fs, dict) else fs
        for f in fs:
            archivos.append(
                dict(
                    pdsid=pid,
                    filename=f.get("FileName"),
                    tipo_archivo=f.get("Type"),
                    tamano_kb=_a_float(f.get("KBytes"), 0.0),
                    url_descarga=f.get("URL"),
                    description=f.get("Description"),
                )
            )
        fila["n_archivos"] = len(fs)
        fila["tamano_kb"] = sum(_a_float(f.get("KBytes"), 0.0) for f in fs)
        meta.append(fila)
    return pd.DataFrame(meta), pd.DataFrame(archivos)


EXT_RASTER = (".img", ".lbl", ".tif", ".tiff")


def choose_raster_product(df_meta, df_files, max_mb=MAX_FILE_MB):
    """One product per AOI: best resolution (lowest m/pixel) within max_mb. Returns (pdsid, files)."""
    if df_files.empty:
        return None, df_files
    f = df_files[df_files["filename"].str.lower().str.endswith(EXT_RASTER, na=False)]
    peso = f.groupby("pdsid")["tamano_kb"].sum() / 1024.0
    ok = peso[peso <= max_mb].index
    if len(ok) == 0:
        return None, f.iloc[0:0]
    m = df_meta[df_meta["pdsid"].isin(ok)].copy()
    m["_peso"] = m["pdsid"].map(peso)
    m["_esc"] = m["escala_m_pix"].fillna(np.inf)
    mejor = m.sort_values(["_esc", "_peso"]).iloc[0]["pdsid"]
    return mejor, f[f["pdsid"] == mejor]


def dir_raw(d: dict) -> Path:
    categoria_dir = {
        "topography": "topography",
        "geology": "geology",
        "mineralogy": "mineralogy",
    }.get(d["category"], d["category"])
    return DIR_RAW / d["body"] / categoria_dir / d["ds_id"]


PRODUCTOS_META = {}  # (ds_id, aoi_id) -> df_meta


def build_download_plan(bodies=BODIES, ids=None, incluir_opcionales=INCLUDE_OPTIONAL_DATASETS):
    filas, resumen = [], []
    for d in DATASET_SELECTION:
        if d["body"] not in bodies or d["access"] == "external":
            continue
        if ids is not None and d["ds_id"] not in ids:
            continue
        if ids is None and not (d["default"] or incluir_opcionales):
            continue
        pts = PLAN_ODE.get(d["ds_id"], [])
        if not pts:
            resumen.append(
                dict(
                    ds_id=d["ds_id"],
                    aoi="-",
                    estado="omitido: no está en ODE (ver paso 3)",
                    n_productos=0,
                    pdsid_elegido="",
                )
            )
            continue
        for aoi in AOIS[d["body"]]:
            bbox = calculate_bounding_box(
                aoi["lat"], aoi["lon"], WINDOW_KM, BODY_RADIUS_KM[d["body"]]
            )
            metas, files = [], []
            for ihid, iid, pt in pts:
                try:
                    m, f = query_ode_products(d["body"], ihid, iid, pt, bbox)
                except Exception as e:
                    resumen.append(
                        dict(
                            ds_id=d["ds_id"],
                            aoi=aoi["id"],
                            estado=f"error {pt}: {e}",
                            n_productos=0,
                            pdsid_elegido="",
                        )
                    )
                    continue
                if not m.empty:
                    m["IHID"], m["IID"], m["PT"] = ihid, iid, pt
                    metas.append(m)
                    files.append(f)
            if not metas:
                resumen.append(
                    dict(
                        ds_id=d["ds_id"],
                        aoi=aoi["id"],
                        estado="sin productos que cubran la AOI",
                        n_productos=0,
                        pdsid_elegido="",
                    )
                )
                continue
            df_m, df_f = pd.concat(metas, ignore_index=True), pd.concat(files, ignore_index=True)
            PRODUCTOS_META[(d["ds_id"], aoi["id"])] = df_m
            elegido = ""
            if d["access"] in ("raster_dem", "raster_map"):
                elegido, sel = choose_raster_product(df_m, df_f)
                elegido = elegido or ""
            elif DOWNLOAD_CUBES:
                sel = df_f[df_f["filename"].str.lower().str.endswith((".lbl",), na=False)].head(5)
            else:
                sel = df_f.iloc[0:0]
            for r in sel.itertuples():
                filas.append(
                    dict(
                        ds_id=d["ds_id"],
                        body=d["body"],
                        aoi=aoi["id"],
                        pdsid=r.pdsid,
                        archivo=r.filename,
                        tamano_mb=round(r.tamano_kb / 1024.0, 2),
                        url=r.url_descarga,
                        destino=str(dir_raw(d) / str(r.filename).lower()),
                    )
                )
            resumen.append(
                dict(
                    ds_id=d["ds_id"],
                    aoi=aoi["id"],
                    estado="ok"
                    if (elegido or d["access"] not in ("raster_dem", "raster_map"))
                    else "no product <= MAX_FILE_MB",
                    n_productos=len(df_m),
                    pdsid_elegido=elegido,
                )
            )
    plan = pd.DataFrame(
        filas, columns=["ds_id", "body", "aoi", "pdsid", "archivo", "size_mb", "url", "destination"]
    )
    return plan, pd.DataFrame(
        resumen
    )  # un producto global puede servir a varias AOI: se deduplica al descargar


try:
    PLAN, RESUMEN_PLAN = build_download_plan()
    PLAN.to_csv(DIR_CATALOGS / "download_plan.csv", index=False, encoding="utf-8-sig")
    display(RESUMEN_PLAN)
    print(
        f"Plan: {len(PLAN)} filas, {PLAN.drop_duplicates('url')['tamano_mb'].sum():,.1f} MB únicos a descargar (tope {MAX_TOTAL_GB} GB)"
    )
except Exception as e:
    PLAN, RESUMEN_PLAN = pd.DataFrame(), pd.DataFrame()
    print("✗ No se pudo construir el plan:", e)
```

## Step 5 — Download to `app/data/raw/{moon,mars}/{topography,geology,mineralogy}/{ds_id}/`
Downloads run only when `DOWNLOAD = True`. Each file is written through a temporary `.part` file, existing files are skipped, and `MAX_TOTAL_GB` is enforced.

```python
def descargar_archivo_con_progreso(url: str, ruta_destino: Path) -> Path:
    ruta_destino = Path(ruta_destino)
    ruta_destino.parent.mkdir(parents=True, exist_ok=True)
    if ruta_destino.exists() and ruta_destino.stat().st_size > 0:
        print(f"→ ya existe: {ruta_destino.name} ({ruta_destino.stat().st_size / 1048576:.2f} MB)")
        return ruta_destino
    tmp = ruta_destino.with_suffix(ruta_destino.suffix + ".part")
    with SESION.get(url, stream=True, timeout=120) as r:
        r.raise_for_status()
        total = int(r.headers.get("content-length", 0))
        with (
            open(tmp, "wb") as f,
            tqdm(
                desc=ruta_destino.name, total=total, unit="B", unit_scale=True, unit_divisor=1024
            ) as pbar,
        ):
            for chunk in r.iter_content(chunk_size=65536):
                if chunk:
                    f.write(chunk)
                    pbar.update(len(chunk))
    tmp.replace(ruta_destino)
    return ruta_destino


def ejecutar_descargas(
    plan: pd.DataFrame, descargar: bool = DOWNLOAD, max_gb: float = MAX_TOTAL_GB
):
    if plan.empty:
        print("Empty plan: nothing to download.")
        return plan
    unicos = plan.drop_duplicates(subset=["url"])
    total_mb = unicos["size_mb"].sum()
    print(f"Plan: {len(unicos)} archivos únicos · {total_mb:,.1f} MB")
    if not descargar:
        print("DRY-RUN (DOWNLOAD=False): nothing was downloaded.")
        return plan
    if total_mb / 1024.0 > max_gb:
        raise RuntimeError(
            f"El plan ({total_mb / 1024:.2f} GB) supera MAX_TOTAL_GB={max_gb}. Ajusta MAX_FILE_MB o el tope."
        )
    estado_url = {}
    for r in unicos.itertuples():
        try:
            descargar_archivo_con_progreso(r.url, Path(r.destino))
            estado_url[r.url] = "ok"
        except Exception as e:
            estado_url[r.url] = f"error: {e}"
    plan = plan.assign(estado=plan["url"].map(estado_url))
    print(pd.Series(estado_url).value_counts().to_string())
    return plan


PLAN = ejecutar_descargas(PLAN)
```

## Step 6 — Process data into `app/data/processed/{moon,mars}/`
The PDS3 reader (`.LBL` label + `.IMG` image) uses `numpy.memmap` to read only each AOI window and uses projection parameters from the label, without requiring GDAL or rasterio.
It supports simple cylindrical and equirectangular projections (MOLA, LOLA, TES, GRS, etc.). Polar products and other formats (GeoTIFF, ISIS cubes) remain in `raw/` and are listed in the log.

```python
from numpy.lib.stride_tricks import sliding_window_view


# ------------------------- lectura de etiquetas PDS3 -------------------------
def leer_etiqueta_pds3(ruta_lbl) -> dict:
    """{objeto: {CLAVE: valor_texto}}; '' = nivel raíz. Primera aparición de cada clave."""
    objetos, pila = {"": {}}, [""]
    for linea in Path(ruta_lbl).read_text(errors="ignore").splitlines():
        line_text = linea.strip()
        if not line_text or line_text.startswith("/*"):
            continue
        if line_text.upper() == "END":
            break
        m = re.match(r"^OBJECT\s*=\s*(\S+)", line_text, re.I)
        if m:
            pila.append(m.group(1).upper())
            objetos.setdefault(pila[-1], {})
            continue
        if re.match(r"^END_OBJECT", line_text, re.I):
            if len(pila) > 1:
                pila.pop()
            continue
        if "=" in line_text:
            k, v = line_text.split("=", 1)
            objetos[pila[-1]].setdefault(k.strip().upper(), v.strip())
    return objetos


def _get(et, clave, objs=("IMAGE", "IMAGE_MAP_PROJECTION", "")):
    for o in objs:
        if o in et and clave in et[o]:
            return et[o][clave]
    for o in et:
        if clave in et[o]:
            return et[o][clave]
    return None


def _num(v, default=None):
    if v is None:
        return default
    m = re.search(r"[-+]?(?:\d+\.?\d*|\.\d+)(?:[eE][-+]?\d+)?", str(v))
    return float(m.group()) if m else default


def _txt(v):
    return None if v is None else str(v).split("<")[0].replace('"', "").strip()


def _buscar_ci(directrio: Path, nombre: str):
    for p in directrio.iterdir():
        if p.name.lower() == nombre.lower():
            return p
    return None


def _puntero_imagen(et, ruta_lbl: Path):
    ptr = et[""].get("^IMAGE") or et.get("IMAGE", {}).get("^IMAGE")
    rec = int(_num(et[""].get("RECORD_BYTES"), 1))
    if ptr is None:
        for ext in (".img", ".IMG"):
            if ruta_lbl.with_suffix(ext).exists():
                return ruta_lbl.with_suffix(ext), 0
        raise FileNotFoundError("Etiqueta sin ^IMAGE y sin .IMG next to ella")
    nombre = re.findall(r'"([^"]+)"', ptr)
    resto = re.sub(r'"[^"]+"', "", ptr)
    n = _num(resto, None)
    en_bytes = "<BYTES>" in ptr.upper()
    off = 0 if n is None else int(n - 1) * (1 if en_bytes else rec)
    if nombre:
        arch = _buscar_ci(ruta_lbl.parent, nombre[0])
        if arch is None:
            raise FileNotFoundError(f"Missing {nombre[0]} next to {ruta_lbl.name}")
        return arch, off
    return ruta_lbl, off


def _dtype_pds(et):
    tipo = (_txt(_get(et, "SAMPLE_TYPE")) or "").upper()
    bits = int(_num(_get(et, "SAMPLE_BITS"), 16))
    if bits == 8:
        return np.dtype("u1" if "UNSIGNED" in tipo or not tipo else "i1")
    little = tipo.startswith(("PC", "LSB", "VAX")) or "LSB" in tipo
    e = "<" if little else ">"
    if "REAL" in tipo or "FLOAT" in tipo:
        return np.dtype(f"{e}f{bits // 8}")
    return np.dtype(f"{e}{'u' if 'UNSIGNED' in tipo else 'i'}{bits // 8}")


def _wrap180(x):
    return (np.asarray(x, dtype=float) + 180.0) % 360.0 - 180.0


def leer_ventana_pds(ruta_lbl, bbox, radio_km_defecto, margen_px=2, banda=0):
    """Read a bounding-box window from a cylindrical PDS3 map. Return scaled values, 1D lat/lon arrays, and pixel dimensions in meters."""
    ruta_lbl = Path(ruta_lbl)
    et = leer_etiqueta_pds3(ruta_lbl)
    proy = (_txt(_get(et, "MAP_PROJECTION_TYPE")) or "").upper()
    if not any(s in proy for s in ("CYLINDRICAL", "EQUIRECTANGULAR")):
        raise NotImplementedError(
            f"Projection '{proy or 'desconocida'}' is unsupported (polar products or other formats)"
        )
    lineas, muestras = int(_num(_get(et, "LINES"))), int(_num(_get(et, "LINE_SAMPLES")))
    nb = int(_num(_get(et, "BANDS"), 1))
    res = _num(_get(et, "MAP_RESOLUTION"))
    lpo, spo = _num(_get(et, "LINE_PROJECTION_OFFSET")), _num(_get(et, "SAMPLE_PROJECTION_OFFSET"))
    clon, clat = _num(_get(et, "CENTER_LONGITUDE"), 0.0), _num(_get(et, "CENTER_LATITUDE"), 0.0)
    radio_km = _num(_get(et, "A_AXIS_RADIUS"), radio_km_defecto)
    if radio_km < 100:
        radio_km = radio_km_defecto
    fac = res * math.cos(math.radians(clat))  # píxeles por grado de longitude

    # filas (i crece hacia el sur):  i = LPO - lat*res
    i0 = int(math.floor(lpo - bbox["maxlat"] * res)) - margen_px
    i1 = int(math.ceil(lpo - bbox["minlat"] * res)) + margen_px
    i0, i1 = max(i0, 0), min(i1, lineas - 1)
    if i1 < i0:
        raise ValueError("AOI outside latitude coverage")
    # columnas:  j = SPO + (lon - clon)*fac
    dc, d = float(_wrap180(bbox["center_lon"] - clon)), bbox["delta_lon_deg"]
    j0 = int(math.floor(spo + (dc - d / 2) * fac)) - margen_px
    j1 = int(math.ceil(spo + (dc + d / 2) * fac)) + margen_px
    js = np.arange(j0, j1 + 1)
    if muestras / fac >= 359.0:  # mapa global: se permite dar la vuelta
        cols = js % muestras
    else:
        cols = js[(js >= 0) & (js < muestras)]
        js = cols
        if cols.size == 0:
            raise ValueError("AOI outside longitude coverage (partial mosaic)")

    arch, off = _puntero_imagen(et, ruta_lbl)
    dt = _dtype_pds(et)
    almacen = (_txt(_get(et, "BAND_STORAGE_TYPE")) or "").upper()
    if nb > 1 and "SEQUENTIAL" not in almacen:
        raise NotImplementedError(f"BAND_STORAGE_TYPE '{almacen}' no soportado")
    if nb > 1:
        mm = np.memmap(arch, dtype=dt, mode="r", offset=off, shape=(nb, lineas, muestras))[banda]
    else:
        mm = np.memmap(arch, dtype=dt, mode="r", offset=off, shape=(lineas, muestras))
    bloque = np.asarray(mm[i0 : i1 + 1])[:, cols].astype(np.float64)

    # valores nulos
    for k in ("MISSING_CONSTANT", "CORE_NULL", "NULL_CONSTANT", "INVALID_CONSTANT"):
        raw = _get(et, k)
        if raw is not None and re.fullmatch(
            r"\s*[-+]?\d+\.?\d*(?:[eE][-+]?\d+)?\s*(<.*>)?\s*", str(raw)
        ):
            bloque[bloque == float(_num(raw))] = np.nan
    bloque[np.abs(bloque) > 1e30] = np.nan

    escala, offset_dn = _num(_get(et, "SCALING_FACTOR"), 1.0), _num(_get(et, "OFFSET"), 0.0)
    valores = bloque * escala + (
        offset_dn if abs(offset_dn) < 500 else 0.0
    )  # OFFSET ~ radio -> alturas relativas

    filas = np.arange(i0, i1 + 1)
    lats = (lpo - filas) / res
    lons = (clon + (js - spo) / fac) % 360.0
    R = radio_km * 1000.0
    dy = R * math.radians(1.0 / res)
    dx = R * np.cos(np.radians(lats)) * math.radians(1.0 / fac)
    return dict(
        valores=valores,
        lats=lats,
        lons=lons,
        dx_m=dx,
        dy_m=dy,
        res_ppd=res,
        pix_m=dy,
        nulo_pct=float(np.isnan(valores).mean() * 100),
        archivo=str(arch),
    )


def normalizar_a_metros(z):
    """Si los valores máximos son < 50 se asume que vienen en km (p. ej. LDEM *_float)."""
    mx = np.nanmax(np.abs(z)) if np.isfinite(z).any() else 0
    if 0 < mx < 50:
        print("  ↳ elevaciones < 50 → se interpretan como km y se pasan a metros")
        return z * 1000.0
    return z


# ------------------------- deriveds topográficas -------------------------


def _ventana(z, k, fun):
    p = k // 2
    w = sliding_window_view(np.pad(z, p, mode="edge"), (k, k))
    return fun(w)


def derivar_topography(z, dx_m, dy_m):
    dz_di, dz_dj = np.gradient(z, axis=0), np.gradient(z, axis=1)
    gN = -dz_di / dy_m  # i crece hacia el sur
    gE = dz_dj / dx_m[:, None]
    pend = np.degrees(np.arctan(np.hypot(gE, gN)))
    asp = np.degrees(np.arctan2(-gE, -gN)) % 360.0
    asp[pend < 0.05] = np.nan
    zp = np.pad(z, 1, mode="edge")
    curv = (zp[1:-1, 2:] - 2 * z + zp[1:-1, :-2]) / (dx_m[:, None] ** 2) + (
        zp[2:, 1:-1] - 2 * z + zp[:-2, 1:-1]
    ) / (dy_m**2)
    rug = _ventana(z, 3, lambda w: np.nanstd(w, axis=(2, 3)))
    rel = _ventana(z, 5, lambda w: np.nanmax(w, axis=(2, 3)) - np.nanmin(w, axis=(2, 3)))
    return pend, asp, rug, curv, rel


def _recortar_bbox(lats, lons, bbox):
    mf = np.abs(lats - bbox["center_lat"]) <= (bbox["maxlat"] - bbox["minlat"]) / 2 + 1e-9
    mc = np.abs(_wrap180(lons - bbox["center_lon"])) <= bbox["delta_lon_deg"] / 2 + 1e-9
    return mf, mc


def process_topography(lbl, bbox, body) -> pd.DataFrame:
    v = leer_ventana_pds(lbl, bbox, BODY_RADIUS_KM[body], margen_px=2)
    z = normalizar_a_metros(v["valores"])
    if v["pix_m"] > WINDOW_KM * 1000 / 5:
        print(
            f"  ⚠ píxel de {v['pix_m']:.0f} m: pocos píxeles en la AOI; busca un producto de mayor resolución"
        )
    pend, asp, rug, curv, rel = derivar_topography(z, v["dx_m"], v["dy_m"])
    mf, mc = _recortar_bbox(v["lats"], v["lons"], bbox)
    LON, LAT = np.meshgrid(v["lons"][mc], v["lats"][mf])
    sl = np.ix_(mf, mc)
    return pd.DataFrame(
        {
            "latitude": LAT.ravel(),
            "longitude": LON.ravel(),
            "elevation_m": z[sl].ravel(),
            "slope_deg": pend[sl].ravel(),
            "aspecto_deg": asp[sl].ravel(),
            "roughness_std3x3_m": rug[sl].ravel(),
            "curvatura_laplaciana_1_m": curv[sl].ravel(),
            "local_relief_5x5_m": rel[sl].ravel(),
        }
    )


def process_raster_map(lbl, bbox, body, variable_name, banda=0) -> pd.DataFrame:
    v = leer_ventana_pds(lbl, bbox, BODY_RADIUS_KM[body], margen_px=0, banda=banda)
    mf, mc = _recortar_bbox(v["lats"], v["lons"], bbox)
    LON, LAT = np.meshgrid(v["lons"][mc], v["lats"][mf])
    return pd.DataFrame(
        {
            "latitude": LAT.ravel(),
            "longitude": LON.ravel(),
            variable_name: v["valores"][np.ix_(mf, mc)].ravel(),
        }
    )


print("✓ Lector PDS3 y deriveds listos")
```

```python
LOG = []


def _log(ds, aoi, accion, estado, detalle=""):
    LOG.append(
        dict(
            ds_id=ds,
            aoi=aoi,
            accion=accion,
            estado=estado,
            detalle=str(detalle)[:300],
            hora=datetime.now().isoformat(timespec="seconds"),
        )
    )


def dir_proc(d):
    return DIR_PROCESSED / d["body"]


def process_all(
    plan: pd.DataFrame, bodies=BODIES, ids=None, incluir_opcionales=INCLUDE_OPTIONAL_DATASETS
):
    """Convert raw data to processed outputs. Rasters become tables; images/cubes become metadata CSVs."""
    por_id = {d["ds_id"]: d for d in DATASET_SELECTION}
    for (ds_id, aoi_id), meta in PRODUCTOS_META.items():
        d = por_id[ds_id]
        if d["body"] not in bodies or (ids is not None and ds_id not in ids):
            continue
        aoi = next(a for a in AOIS[d["body"]] if a["id"] == aoi_id)
        bbox = calculate_bounding_box(aoi["lat"], aoi["lon"], WINDOW_KM, BODY_RADIUS_KM[d["body"]])
        sufijo = f"{ds_id}__{aoi_id}"
        # metadata (siempre)
        meta.assign(ds_id=ds_id, aoi=aoi_id).to_csv(
            dir_proc(d) / f"{sufijo}__metadatas.csv", index=False, encoding="utf-8-sig"
        )
        _log(ds_id, aoi_id, "metadata", "ok", f"{len(meta)} productos")
        if d["access"] not in ("raster_dem", "raster_map"):
            continue
        # raster: localizar .lbl descargado
        sub = plan[(plan.ds_id == ds_id) & (plan.aoi == aoi_id)] if not plan.empty else plan
        lbls = (
            [
                Path(x)
                for x in sub["destination"]
                if str(x).lower().endswith(".lbl") and Path(x).exists()
            ]
            if not sub.empty
            else []
        )
        if not lbls:
            _log(ds_id, aoi_id, "raster", "omitido", "sin .LBL descargado (¿DOWNLOAD=False?)")
            continue
        try:
            if d["access"] == "raster_dem":
                df = process_topography(lbls[0], bbox, d["body"])
            else:
                df = process_raster_map(lbls[0], bbox, d["body"], variable_name=f"{ds_id}_banda0")
            df.to_csv(dir_proc(d) / f"{sufijo}.csv", index=False)
            _log(ds_id, aoi_id, "raster", "ok", f"{len(df)} filas")
            print(f"✓ {sufijo}: {len(df):,} filas")
        except Exception as e:
            _log(ds_id, aoi_id, "raster", "error", e)
            print(f"✗ {sufijo}: {e}")


process_all(PLAN)
df_log = pd.DataFrame(LOG)
display(df_log if not df_log.empty else "Sin acciones (¿dry-run?)")
```

## Step 7 — Shared grid (`GRID_KM`) for the three categories
Aggregate each pixel table by grid cell (mean; circular mean for aspect). Optional layers include external geology (spatial join) and crater density when paths are set in `GEOLOGY_VECTOR` / `CRATER_CATALOG`.

```python
def assign_grid_cell(lat, lon, bbox, malla_km):
    """Cell indices (i = east, j = north) relative to the AOI southwest corner."""
    R = bbox["radio_km"]
    lat0 = bbox["center_lat"] - (bbox["maxlat"] - bbox["minlat"]) / 2
    cosc = math.cos(math.radians(bbox["center_lat"]))
    y = R * np.radians(np.asarray(lat, dtype=float) - lat0)
    x = (
        R
        * cosc
        * np.radians(
            _wrap180(np.asarray(lon, dtype=float) - bbox["center_lon"]) + bbox["delta_lon_deg"] / 2
        )
    )
    return np.floor(x / malla_km).astype(int), np.floor(y / malla_km).astype(int)


def _media_circular(s):
    r = np.radians(s.dropna())
    return (
        np.nan
        if r.empty
        else float(np.degrees(np.arctan2(np.sin(r).mean(), np.cos(r).mean())) % 360.0)
    )


def agregar_a_malla(df, bbox, malla_km):
    ci, cj = assign_grid_cell(df["latitude"].values, df["longitude"].values, bbox, malla_km)
    t = df.drop(columns=["latitude", "longitude"]).assign(celda_i=ci, celda_j=cj)
    agg = {
        c: (_media_circular if c.startswith("aspect") else "mean")
        for c in t.columns
        if c not in ("cell_i", "cell_j")
    }
    out = t.groupby(["cell_i", "cell_j"]).agg(agg)
    out["n_pixeles"] = t.groupby(["cell_i", "cell_j"]).size()
    return out


def centroides(celdas, bbox, malla_km):
    R = bbox["radio_km"]
    lat0 = bbox["center_lat"] - (bbox["maxlat"] - bbox["minlat"]) / 2
    lon_w = bbox["center_lon"] - bbox["delta_lon_deg"] / 2
    cosc = math.cos(math.radians(bbox["center_lat"]))
    lat = lat0 + np.degrees((celdas["cell_j"] + 0.5) * malla_km / R)
    lon = (lon_w + np.degrees((celdas["cell_i"] + 0.5) * malla_km / (R * cosc))) % 360.0
    return lat, lon


def join_vector_geology(tab, body, bbox):
    cfg = GEOLOGY_VECTOR[body]
    if not cfg["ruta"]:
        return tab
    try:
        import geopandas as gpd
        from shapely.geometry import Point

        g = gpd.read_file(cfg["ruta"])
        usa_neg = g.total_bounds[0] < 0
        pts = gpd.GeoDataFrame(
            tab[["cell_i", "cell_j"]].copy(),
            geometry=[
                Point(((lo + 180) % 360 - 180) if usa_neg else lo, la)
                for la, lo in zip(tab.latitude, tab.longitude, strict=True)
            ],
            crs=g.crs,
        )
        cols = [c for c in (cfg["col_unidad"], cfg["col_edad"]) if c]
        j = gpd.sjoin(pts, g[cols + ["geometry"]], how="left", predicate="within").drop_duplicates(
            ["cell_i", "cell_j"]
        )
        return tab.merge(
            j[["cell_i", "cell_j"] + cols].rename(
                columns={
                    cfg["col_unidad"]: "geologic_unit",
                    **({cfg["col_edad"]: "estimated_age"} if cfg["col_edad"] else {}),
                }
            ),
            on=["cell_i", "cell_j"],
            how="left",
        )
    except Exception as e:
        print("  ⚠ vector geology was not applied:", e)
        return tab


def join_craters(tab, body, bbox, malla_km):
    cfg = CRATER_CATALOG[body]
    if not cfg["ruta"]:
        return tab
    try:
        c = pd.read_csv(
            cfg["ruta"],
            usecols=[
                x for x in (cfg["col_lat"], cfg["col_lon"], cfg["col_diam"], cfg["col_prof"]) if x
            ],
        )
        c = c.rename(
            columns={
                cfg["col_lat"]: "latitude",
                cfg["col_lon"]: "longitude",
                cfg["col_diam"]: "diam_km",
            }
        )
        c["longitude"] = c["longitude"] % 360.0
        ci, cj = assign_grid_cell(c["latitude"].values, c["longitude"].values, bbox, malla_km)
        c["cell_i"], c["cell_j"] = ci, cj
        c = c[(c.celda_i >= 0) & (c.celda_j >= 0)]
        g = (
            c.groupby(["cell_i", "cell_j"])
            .agg(n_crateres=("diam_km", "size"), diametro_crater_medio_km=("diam_km", "mean"))
            .reset_index()
        )
        if cfg["col_prof"]:
            g = g.merge(
                c.groupby(["cell_i", "cell_j"])[cfg["col_prof"]]
                .mean()
                .rename("mean_crater_depth_km")
                .reset_index(),
                on=["cell_i", "cell_j"],
                how="left",
            )
        g["crater_density_km2"] = g["crater_count"] / (malla_km**2)
        return tab.merge(g, on=["cell_i", "cell_j"], how="left").fillna(
            {"crater_count": 0, "crater_density_km2": 0.0}
        )
    except Exception as e:
        print("  ⚠ crater catalog was not applied:", e)
        return tab


def build_integrated_grid(body, aoi, malla_km=GRID_KM):
    bbox = calculate_bounding_box(aoi["lat"], aoi["lon"], WINDOW_KM, BODY_RADIUS_KM[body])
    tablas = []
    for f in sorted((DIR_PROCESSED / body).glob(f"*__{aoi['id']}.csv")):
        if f.name.startswith("INTEGRATED"):
            continue
        df = pd.read_csv(f)
        if {"latitude", "longitude"} <= set(df.columns):
            t = agregar_a_malla(df, bbox, malla_km)
            tablas.append(t)
    if not tablas:
        print(f"  (sin tablas procesadas para {aoi['id']})")
        return None
    tab = pd.concat(tablas, axis=1).reset_index()
    tab = tab.loc[:, ~tab.columns.duplicated()]
    tab["latitude"], tab["longitude"] = centroides(tab, bbox, malla_km)
    tab = join_vector_geology(tab, body, bbox)
    tab = join_craters(tab, body, bbox, malla_km)
    cols = ["latitude", "longitude"] + [
        c for c in tab.columns if c not in ("latitude", "longitude")
    ]
    tab = tab[cols]
    tab.to_csv(DIR_PROCESSED / body / f"INTEGRATED__{aoi['id']}.csv", index=False)
    print(f"✓ INTEGRATED__{aoi['id']}: {len(tab)} celdas de {malla_km} km, {tab.shape[1]} columnas")
    return tab


MALLAS = {}
for body in BODIES:
    for aoi in AOIS[body]:
        MALLAS[aoi["id"]] = build_integrated_grid(body, aoi)
```

## Step 8 — Final dataset catalog
Scan `processed/` CSVs and `raw/` files and report row counts, columns, valid values, null percentage, geographic range, and file size.

```python
def _resumen_csv(ruta: Path) -> dict:
    filas, nulos, cols = 0, 0, None
    lat_min = lat_max = lon_min = lon_max = np.nan
    celdas = 0
    for ch in pd.read_csv(ruta, chunksize=200_000, low_memory=False):
        cols = list(ch.columns) if cols is None else cols
        filas += len(ch)
        nulos += int(ch.isna().sum().sum())
        celdas += ch.size
        if "latitude" in ch:
            lat_min, lat_max = (
                np.nanmin([lat_min, ch.latitude.min()]),
                np.nanmax([lat_max, ch.latitude.max()]),
            )
        if "longitude" in ch:
            lon_min, lon_max = (
                np.nanmin([lon_min, ch.longitude.min()]),
                np.nanmax([lon_max, ch.longitude.max()]),
            )
    return dict(
        filas=filas,
        columnas=len(cols or []),
        lista_columnas=", ".join(cols or []),
        valores_validos=celdas - nulos,
        pct_nulos=round(100 * nulos / celdas, 2) if celdas else np.nan,
        lat_min=lat_min,
        lat_max=lat_max,
        lon_min=lon_min,
        lon_max=lon_max,
    )


def construir_catalogo_final() -> pd.DataFrame:
    por_id = {d["ds_id"]: d for d in DATASET_SELECTION}
    filas = []
    for body in BODIES:
        for f in sorted((DIR_PROCESSED / body).glob("*.csv")):
            partes = f.stem.split("__")
            ds_id, aoi = partes[0], (partes[1] if len(partes) > 1 else "")
            tipo = (
                partes[2]
                if len(partes) > 2
                else ("integrado" if ds_id == "INTEGRATED" else "tabla")
            )
            d = por_id.get(ds_id, {})
            filas.append(
                dict(
                    nivel="processed",
                    body=body,
                    ds_id=ds_id,
                    category=d.get("category", "todas" if ds_id == "INTEGRATED" else ""),
                    instrument=d.get(
                        "instrument", "malla integrada" if ds_id == "INTEGRATED" else ""
                    ),
                    aoi=aoi,
                    tipo=tipo,
                    archivo=str(f.relative_to(DATA_DIR)),
                    formato="csv",
                    tamano_mb=round(f.stat().st_size / 1048576, 3),
                    **_resumen_csv(f),
                )
            )
        for f in sorted((DIR_RAW / body).rglob("*")):
            if f.is_file() and not f.name.endswith(".part"):
                ds_id = f.parent.name
                d = por_id.get(ds_id, {})
                filas.append(
                    dict(
                        nivel="raw",
                        body=body,
                        ds_id=ds_id,
                        category=d.get("category", ""),
                        instrument=d.get("instrument", ""),
                        aoi="",
                        tipo="crudo",
                        archivo=str(f.relative_to(DATA_DIR)),
                        formato=f.suffix.lower().strip("."),
                        tamano_mb=round(f.stat().st_size / 1048576, 3),
                    )
                )
    cat = pd.DataFrame(filas)
    return cat


df_catalogo_final = construir_catalogo_final()
ruta = DIR_CATALOGS / "final_dataset_catalog.csv"
df_catalogo_final.to_csv(ruta, index=False, encoding="utf-8-sig")
print(f"✓ Final catalog: {len(df_catalogo_final)} archivos → {ruta.name}")
if not df_catalogo_final.empty:
    proc = df_catalogo_final.query("nivel == 'processed'")
    display(
        proc.groupby(["body", "category"], dropna=False).agg(
            archivos=("archivo", "count"),
            filas=("filas", "sum"),
            valores_validos=("valores_validos", "sum"),
            mb=("size_mb", "sum"),
        )
    )
    display(df_catalogo_final.drop(columns=["lista_columnas"], errors="ignore"))
```
